```python
# Find the first non-repeating character
from collections import Counter
s = "aabbcde"
for c in s:
    if Counter(s)[c] == 1:
        print(c)  # 'c'
        break
```

```python
# find all the non-repeating characters
 # simpplfy it 
non_repeating = [c for c in s if Counter(s)[c] == 1]
print(non_repeating)  # ['c', 'd', 'e']
```
