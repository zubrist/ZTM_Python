> In Python, variables are symbolic names that refer to objects or values stored in your computer’s memory.

- We define variables by assigning them a value using the assignment operator.
- Python variables are dynamically typed, allowing type changes through reassignment.
- Python variable names can include letters, digits, and underscores but can’t start with a digit. We should use snake case for multi-word names to improve readability.
- Variables exist in different scopes (global, local, non-local, or built-in), which affects how we can access them.
- We can have an unlimited number of variables in Python, limited only by computer memory.

**Syntax for Variable Assignment:**
```python
variable_name = value
```

````python
contacts = [
     ("Joe", "111-2222-3333", "joe@example.com"),
     ("Lara", "111-2222-3333", "lara@example.com"),
     ("Linda", "111-2222-3333", "linda@example.com"),
     ("David", "111-2222-3333", "david@example.com"),
     ("Jane", "111-2222-3333", "jane@example.com"),
 ]

for name, phone, email in contacts:
    print(f"Name: {name}, Phone: {phone}, Email: {email}")
````

## Public and Non-Public Variable Names
A widely used naming convention for variables in Python is to use a leading underscore when you need to communicate that a given variable is what Python defines as non-public. A non-public variable is a variable that shouldn’t be used outside its defining module. These variables are for internal use only:

```python
_timeout = 30

def get_timeout():
    return _timeout

def set_timeout(seconds):
    global _timeout
    _timeout = seconds
print(get_timeout())  # Output: 30
set_timeout(60)
print(get_timeout())  # Output: 60
``` 

## Variables Hold References to Objects
What happens when you create a variable with an assignment? The variable holds a reference to the object in memory. When you assign a new value to the variable, it points to a new object, and the old object may be garbage collected if there are no more references to it.

```python
>>> 73
73
```
Python does the following operations:
- Creates an integer object with the value 73 in memory.
- Displays the value of the object (73) in the output.

We can see an integer object is created using the built-in type() function:
```python
>>> type(73)
<class 'int'>
``` 

> A Python variable is a symbolic name that refers to or points to an object like 73. Once an object is assigned to a variable, you can refer to it by the variable’s name, but the data itself is still contained within the object.

For example :
```python
x = 73
print(x)  # Output: 73
```
