## Types of errros in Python
There are several types of errors in Python, including:
1. **Syntax Errors**: These occur when the code is not written in a valid Python syntax. For example, missing a colon at the end of a function definition or using incorrect indentation.
2. **Runtime Errors**: These occur during the execution of the program. For example,
    - **ZeroDivisionError**: When you try to divide a number by zero.
    - **NameError**: When you try to use a variable that has not been defined.
    - **TypeError**: When you try to perform an operation on incompatible data types.
3. **Logical Errors**: These occur when the code runs without throwing any errors, but produces incorrect results. For example, using the wrong operator or incorrect logic in a conditional statement.
4. **Import Errors**: These occur when you try to import a module that does not exist or is not installed in your environment.
5. **File Not Found Errors**: These occur when you try to access a file that does not exist in the specified location.
6. **Value Errors**: These occur when a function receives an argument of the correct type but an inappropriate value. For example, trying to convert a string that does not represent a number into an integer.
7. **Index Errors**: These occur when you try to access an index that is out of range in a list or other sequence.
8. **Key Errors**: These occur when you try to access a key that does not exist in a dictionary.
9. **Attribute Errors**: These occur when you try to access an attribute that does not exist for an object.
10. **Indentation Errors**: These occur when the code is not properly indented, which is crucial in Python for defining code blocks.

 ## Types of error handling
There are several types of error handling techniques that can be used in programming, including:
1. **Try-Catch Blocks**: This is a common method where you wrap code that may throw an error in a try block, and handle the error in a catch block.


```python   
try:
    # Code that may throw an error
    result = 10 / 0
except ZeroDivisionError as e:
    print("Cannot divide by zero:", e)
```
2. **Finally Block**: This block is used to execute code that must run regardless of whether an error occurred or not. It is often used for cleanup actions.

```python
try:
    # Code that may throw an error
    result = 10 / 0
except ZeroDivisionError as e:
    print("Cannot divide by zero:", e)
finally:
    print("This block always executes.")
```
3. **Raising Exceptions**: You can raise exceptions intentionally when a certain condition is met in your code.

```python
def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    return a / b    
try:    
    result = divide(10, 0)
except ValueError as e:
    print(e)
```
4. **Custom Exceptions**: You can create your own exception classes to handle specific error conditions in a more descriptive way.

```python
class CustomError(Exception):
    pass

def divide(a, b):
    if b == 0:
        raise CustomError("Cannot divide by zero.")
    return a / b

try:
    result = divide(10, 0)
except CustomError as e:
    print(e)
```
5. **Using Assertions**: Assertions are used to check for conditions that should always be true in your code. If the condition is false, an AssertionError is raised.

```python
def divide(a, b):
    assert b != 0, "Cannot divide by zero."
    return a / b

try:
    result = divide(10, 0)
except AssertionError as e:
    print(e)
```
6. **Context Managers**: These are used to manage resources, such as file handling, ensuring that resources are properly released after use.
```python
with open('file.txt', 'r') as file:
    data = file.read()
```
In this example, the file will be automatically closed after the block of code is executed, even if an error occurs.

7. **Logging**: Instead of just printing error messages, you can use logging to record errors in a more structured way, which can be helpful for debugging and monitoring applications.

```python
import logging
logging.basicConfig(filename='app.log', level=logging.ERROR)
try:
    result = 10 / 0
except ZeroDivisionError as e:
    logging.error("Cannot divide by zero: %s", e)
``` 



## understanding Loggin for all the levels and when and how to use it
Logging is a powerful tool for tracking events that happen when some software runs. It can be used to record information about the execution of a program, which can be helpful for debugging and monitoring applications. Python's built-in logging module provides several levels of logging, each with its own purpose:
1. **DEBUG**: This level is used for detailed information, typically of interest only when diagnosing problems. It is the lowest level of logging and is usually turned off in production environments.

```python   
import logging
logging.basicConfig(level=logging.DEBUG)

#inside a function or method
logging.debug("This is a debug message.")
```

```python
import logging

logging.basicConfig(level=logging.WARNING, format="%(levelname)s: %(message)s")

def process_payment(amount: float, balance: float):
    if balance < 1000:
        logging.warning("Account balance is low: %.2f", balance)  # still works, but risky soon
    
    if amount > balance:
        raise ValueError("Insufficient balance")

    return "Payment successful"

print(process_payment(200, 900))
``` 

4. **ERROR**: This level is used to indicate a more serious problem that has occurred. The software has not been able to perform some function.

```python
import logging
logging.basicConfig(level=logging.ERROR, format = "%(levelname)s : %(asctime)s: #%(filename)s: %(lineno)d %(message)s")


def payment_processing(amount: float, balance: float):
    if amount > balance:
        logging.error("Payment failed due to insufficient balance. Attempted amount: %.2f, Available balance: %.2f", amount, balance)
        raise ValueError("Insufficient balance")
    return "Payment successful"
```    


>  Difference between logging an error and raising an exception

***Key difference:***

```logging.error()``` — records the error but continues execution.

```raise ValueError()``` — stops execution and signals an error to the caller.


5. **CRITICAL**: This level is used to indicate a very serious error that has occurred. The software may be unable to continue running.
    
```python
import logging
logging.basicConfig(level=logging.CRITICAL, format = "%(levelname)s : %(asctime)s: #%(filename)s: %(lineno)d %(message)s")

def critical_error_example():
    logging.critical("A critical error occurred. The system may be unable to continue.")
    raise SystemExit("Critical error encountered. Exiting program.")
```     


## HTTP Status Codes and what they mean and when to use them
HTTP status codes are standardized codes that indicate the result of an HTTP request. They are categorized into five classes based on the first digit of the code:
1. **1xx (Informational)**: These codes indicate that the request has been received and is being processed. They are rarely used in practice.
    - Example: 100 Continue, 101 Switching Protocols    
2. **2xx (Successful)**: These codes indicate that the request was successfully received, understood, and accepted.
    - Example: 200 OK, 201 Created, 204 No Content
3. **3xx (Redirection)**: These codes indicate that further action is needed to complete the request, usually in the form of a redirection.
    - Example: 301 Moved Permanently, 302 Found, 304 Not Modified
4. **4xx (Client Error)**: These codes indicate that the client made an error in the request. The server cannot process the request due to a client error.
    - Example: 400 Bad Request, 401 Unauthorized, 403 Forbidden, 404 Not Found
5. **5xx (Server Error)**: These codes indicate that the server encountered an error while processing the request. The server is aware that it has encountered an error or is otherwise incapable of performing the request.
    - Example: 500 Internal Server Error, 502 Bad Gateway, 503 Service Unavailable  
### When to use HTTP status codes:
- Use **2xx** codes when the request is successful and the server has fulfilled the request.
- Use **3xx** codes when the client needs to take additional action to complete the request, such as following a redirect.
- Use **4xx** codes when the client has made an error in the request, such as providing invalid input or trying to access a resource that does not exist.
- Use **5xx** codes when the server has encountered an error while processing the request, such as a database failure or an unexpected condition that prevents it from fulfilling the request.  


### list of 2xx status codes and their meaning in FastAPI:
- **200 OK**: The request has succeeded. The meaning of the success depends on the HTTP method used (GET, POST, etc.).
- **201 Created**: The request has been fulfilled and has resulted in the creation of a new resource. This is typically used in response to POST requests.
- **202 Accepted**: The request has been accepted for processing, but the processing has not been completed. This is typically used for asynchronous processing.
- **204 No Content**: The server successfully processed the request, but is not returning any content. This is typically used in response to DELETE requests or when the server has fulfilled the request but does not need to return any content.
- **205 Reset Content**: The server successfully processed the request, but is not returning any    content and requires the requester to reset the document view. This is typically used when the server wants the client to clear the form or reset the user interface.
- **206 Partial Content**: The server is delivering only part of the resource due to a range header sent by the client. This is typically used for resuming interrupted downloads or for streaming media content.       
- **207 Multi-Status**: The message body that follows is an XML message and can contain a number of separate response codes, depending on how many sub-requests were made. This is typically used in WebDAV applications.
- **208 Already Reported**: The members of a DAV binding have already been enumerated in a previous reply to this request, and are not being included again. This is typically used in WebDAV applications.
- **226 IM Used**: The server has fulfilled a GET request for the resource, and the response is a representation of the result of one or more instance-manipulations applied to the current instance. This is typically used in HTTP Delta encoding.    
### list of 4xx status codes and their meaning in FastAPI:
- **400 Bad Request**: The server cannot or will not process the request due to a client error (e.g., malformed request syntax, invalid request message framing, or deceptive request routing).
- **401 Unauthorized**: The request has not been applied because it lacks valid authentication credentials for the target resource. This is typically used when authentication is required and has failed or has not yet been provided.
- **403 Forbidden**: The server understood the request but refuses to authorize it. This is typically used when the client does not have permission to access the requested resource.
- **404 Not Found**: The server can't find the requested resource. This is typically used when the requested resource does not exist on the server.
- **405 Method Not Allowed**: The request method is known by the server but is not supported by the target resource. This is typically used when the client tries to use an HTTP method that is not allowed for the requested resource.
- **406 Not Acceptable**: The server cannot produce a response matching the list of acceptable values defined in the request's proactive content negotiation headers, and the server is unwilling to supply a default representation. This is typically used when the client requests a specific content type that the server cannot provide.
- **407 Proxy Authentication Required**: The client must first authenticate itself with the proxy. This is typically used when the client needs to authenticate with a proxy server before accessing the requested resource.
- **408 Request Timeout**: The server did not receive a complete request message within the time that it was prepared to wait. This is typically used when the client takes too long to send a request or when the server is under heavy load.
- **409 Conflict**: The request could not be completed due to a conflict with the current state of the target resource. This is typically used when there is a conflict in the request, such as an edit conflict in a version control system.
- **410 Gone**: The target resource is no longer available at the origin server and that this condition is likely to be permanent. This is typically used when a resource has been intentionally removed and the server wants to indicate that it should not be requested again.
- **411 Length Required**: The server refuses to accept the request without a defined Content-Length. This is typically used when the client fails to provide a Content-Length header in a request that requires it.
- **412 Precondition Failed**: One or more conditions given in the request header fields evaluated
to false when tested on the server. This is typically used when the client provides preconditions in the request headers that are not met by the server.
- **413 Payload Too Large**: The server is refusing to process a request because the request payload is larger than the server is willing or able to process. This is typically used when the client sends a request with a body that exceeds the server's limits.
- **414 URI Too Long**: The server is refusing to service the request because the request-target is longer than the server is willing to interpret. This is typically used when the client sends a request with a URI that exceeds the server's limits.
- **415 Unsupported Media Type**: The server is refusing to service the request because the payload format is not supported by the server for the target resource. This is typically used when the client sends a request with a content type that the server does not support.
- **416 Range Not Satisfiable**: The server cannot serve the requested byte range.
This is typically used when the client requests a range of bytes that is not available in the resource.
- **417 Expectation Failed**: The expectation given in the request's Expect header field could not be met by the server. This is typically used when the client sends a request with an Expect header that the server cannot fulfill

### list of 5xx status codes and their meaning in FastAPI:
- **500 Internal Server Error**: The server encountered an unexpected condition that prevented it from fulfilling the request. This is a generic error message when no more specific message is suitable.
- **501 Not Implemented**: The server does not support the functionality required to fulfill the
request. This is typically used when the server does not recognize the request method or lacks the ability to fulfill the request.
- **502 Bad Gateway**: The server, while acting as a gateway or proxy, received an invalid response from the upstream server it accessed in attempting to fulfill the request. This is typically used when the server is acting as a proxy and receives an invalid response from the upstream server.
- **503 Service Unavailable**: The server is currently unable to handle the request due to temporary overload or scheduled maintenance, which will likely be alleviated after some delay. This is typically used when the server is temporarily unavailable due to high load or maintenance.
- **504 Gateway Timeout**: The server, while acting as a gateway or proxy, did not receive a timely response from the upstream server specified by the URI (e.g., HTTP, FTP, LDAP) or some other auxiliary server (e.g., DNS) it needed to access in attempting to fulfill the request. This is typically used when the server is acting as a proxy and does not receive a response from the upstream server within a certain time frame.
- **505 HTTP Version Not Supported**: The server does not support the HTTP protocol version that was used in the request. This is typically used when the client sends a request with an HTTP version that the server does not support.
- **506 Variant Also Negotiates**: The server has an internal configuration error: the chosen variant resource is configured to engage in content negotiation itself, and is therefore not a proper end point in the negotiation process. This is typically used when there is a configuration error on the server related to content negotiation.
- **507 Insufficient Storage**: The server is unable to store the representation needed to complete the request. This is typically used when the server is unable to store the requested resource due to insufficient storage space.
- **508 Loop Detected**: The server detected an infinite loop while processing the request. This is typically used when the server encounters an infinite loop while processing a request, such as a recursive call that does not have a base case.
- **510 Not Extended**: Further extensions to the request are required for the server to fulfill it. This is typically used when the server requires additional extensions to the request in order to fulfill it.
